```python
numeric_columns = df.select_dtypes(include=np.number).columns

print(numeric_columns)
```

STEP 11 — Handle Missing Numerical Values

```python
for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())
```

STEP 12 — Handle Missing Categorical Values

```python
categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])
```

# PART 2 — OUTLIER DETECTION

STEP 14 — Select Suitable Numerical Columns

For this dataset, we'll focus on meaningful continuous/count variables rather than IDs.

```python
outlier_columns = [
    "unit_price",
    "quantity",
    "discount_percent",
    "discount_amount",
    "revenue",
    "pages_viewed",
    "time_on_site_sec",
    "rating",
    "review_helpful_votes"
]
```

STEP 15 — Calculate Q1

```python
Q1 = df[outlier_columns].quantile(0.25)

print(Q1)
```

STEP 16 — Calculate Q3

```python
Q3 = df[outlier_columns].quantile(0.75)

print(Q3)
```

STEP 17 — Calculate IQR

```python
IQR = Q3 - Q1

print(IQR)
```

STEP 18 — Calculate Lower and Upper Limits

```python
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:")
print(lower_limit)

print("\nUpper Limit:")
print(upper_limit)
```

STEP 19 — Count Outliers

```python
outlier_count = (
    (df[outlier_columns] < lower_limit) |
    (df[outlier_columns] > upper_limit)
).sum()

print(outlier_count)
```

STEP 20 — Visualize Outliers Before Handling

```python
plt.figure(figsize=(15, 7))

df[outlier_columns].boxplot()

plt.title("Outliers Before Handling")
plt.xticks(rotation=45)
plt.show()
```

STEP 21 — Neutralize Outliers

```python
for column in outlier_columns:
    df[column] = df[column].clip(
        lower=lower_limit[column],
        upper=upper_limit[column]
    )
```

STEP 22 — Check Boxplot After Handling

```python
plt.figure(figsize=(15, 7))

df[outlier_columns].boxplot()

plt.title("Outliers After Handling")
plt.xticks(rotation=45)
plt.show()
```

# PART 3 — FEATURE ENGINEERING

Feature 1 — Total Price Before Discount

```python
df["total_price"] = df["unit_price"] * df["quantity"]
```

Feature 2 — Effective Price After Discount

```python
df["effective_price"] = (
    df["unit_price"] *
    (1 - df["discount_percent"] / 100)
)
```

Feature 3 — Revenue Per Page

```python
df["revenue_per_page"] = (
    df["revenue"] / (df["pages_viewed"] + 1)
)
```

STEP 23 — Create More Features

Feature 4 — Revenue Per Second

```python
df["revenue_per_second"] = (
    df["revenue"] / (df["time_on_site_sec"] + 1)
)
```

Feature 5 — Discount Rate

```python
df["discount_rate"] = df["discount_percent"] / 100
```

Feature 6 — Engagement Score

```python
df["engagement_score"] = (
    df["pages_viewed"] *
    df["time_on_site_sec"]
)
```
